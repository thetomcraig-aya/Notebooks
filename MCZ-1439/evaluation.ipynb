# MCZ-1439 CSV Load

Load a CSV file from this folder into a pandas dataframe.

```python
from pathlib import Path

import pandas as pd

folder = Path('/Users/tomcraig/dev/mcz_notebooks_and_queries/MCZ-1439')
csv_files = sorted(folder.glob('*.csv'))
csv_files
```

```python
csv_path = folder / "example_input_clean.csv"
df = pd.read_csv(csv_path, dtype=str, keep_default_na=False)

def clean_whitespace(value):
    if isinstance(value, str):
        value = " ".join(value.split())
        return value.replace("[ ", "[").replace(" ]", "]")
    return value

df = df.apply(lambda column: column.map(clean_whitespace) if column.dtype == "object" else column)

clean_csv_path = folder / "example_input_cleaned.csv"
df.to_csv(clean_csv_path, index=False)
print(f"Wrote {clean_csv_path}")

df.head()
```
