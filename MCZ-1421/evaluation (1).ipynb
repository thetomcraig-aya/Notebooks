```python
# Find taxonomy codes from the JSON that exist in df_nucc_lookup.code
json_keys_set = set(keys)
nucc_codes = set(df_nucc_lookup['Code'].astype(str))

present_in_lookup = df_nucc_lookup[df_nucc_lookup['Code'].astype(str).isin(json_keys_set)]
missing_in_lookup = sorted(json_keys_set - nucc_codes)

print(f"Codes found in df_nucc_lookup: {len(present_in_lookup)}")
print(present_in_lookup[['Code']])
print(f"\nCodes missing from df_nucc_lookup: {len(missing_in_lookup)}")
if missing_in_lookup:
    print(missing_in_lookup)
else:
    print("None")
```

```python
import json
import pandas as pd

# Build updated JSON using codes present in both the taxonomy file and df_nucc_lookup
code_col = 'Code' if 'Code' in df_nucc_lookup.columns else 'code'
classification_col = 'Classification' if 'Classification' in df_nucc_lookup.columns else 'classification'
specialization_col = 'Specialization' if 'Specialization' in df_nucc_lookup.columns else 'specialization'

json_keys_set = set(unique_taxonomy_codes)
nucc_filtered = df_nucc_lookup[df_nucc_lookup[code_col].astype(str).isin(json_keys_set)]

# Helper to clean string values
clean = lambda val: "" if pd.isna(val) else str(val).strip()

output_json = {}
for _, row in nucc_filtered.iterrows():
    code_val = str(row[code_col]).strip()
    output_json[code_val] = {
        "taxonomy_code": code_val,
        "role": "",
        "specialty": clean(row[classification_col]),
        "sub_specialty": clean(row[specialization_col])
    }

print(json.dumps(output_json, indent=2))
```

```python
import json

# Read the file with utf-8-sig to handle BOM
with open('/Users/tomcraig/dev/ETL/WinnowPackages/Aya.Winnow.Domain.Model/Taxonomies/taxonomycodes.json', 'r', encoding='utf-8-sig') as f:
    content = f.read()

# Load as JSON
updated_taxonomy = json.loads(content)
print(f"Loaded {len(updated_taxonomy)} entries from taxonomycodes.json")
print(json.dumps(updated_taxonomy, indent=2))
```

```python
# Fill empty sub_specialty with specialty value
original_count = len(updated_taxonomy)
print(f"Starting with {original_count} entries")

for code, entry in updated_taxonomy.items():
    if not entry.get("sub_specialty") or entry["sub_specialty"].strip() == "":
        entry["sub_specialty"] = entry.get("specialty", "")

final_count = len(updated_taxonomy)
print(f"After update: {final_count} entries")
print(f"Entries preserved: {final_count == original_count}")

print(json.dumps(updated_taxonomy, indent=2))

```
