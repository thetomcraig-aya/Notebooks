```python
from pathlib import Path

import pandas as pd

base_dir = Path('/Users/tomcraig/dev/mcz_notebooks_and_queries/WIN-doccafe active')
dataframes = {}
load_notes = {}

for csv_path in sorted(base_dir.glob('*.csv')):
    df_name = csv_path.stem.lower().replace(' ', '_').replace('(', '').replace(')', '').replace('-', '_') + '_df'
    try:
        df = pd.read_csv(csv_path)
        load_notes[df_name] = 'loaded with default parser'
    except pd.errors.ParserError:
        df = pd.read_csv(csv_path, engine='python', on_bad_lines='skip')
        load_notes[df_name] = 'loaded with python engine; skipped malformed rows'
    dataframes[df_name] = df
    globals()[df_name] = df

for df_name, df in dataframes.items():
    print(f"{df_name} ({len(df.columns)} columns)")
    print(load_notes[df_name])
    print(list(df.columns))
    print()
```
