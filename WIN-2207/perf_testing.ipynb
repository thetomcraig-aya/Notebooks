# Load Files

```python
import pandas as pd
from pathlib import Path

DATA_DIR = Path("/Users/tomcraig/dev/mcz_notebooks_and_queries/WIN-2207/precompute_output_example_files/20260316220105")

csv_files = sorted(DATA_DIR.glob("*-domain_merged.csv"))

combined_df = pd.concat(
    [pd.read_csv(csv_file, low_memory=False).assign(source_file=csv_file.name) for csv_file in csv_files],
    ignore_index=True,
)

print(f"Loaded {len(csv_files)} files into one DataFrame")
# print(combined_df.head())


```

```python
distinct_cand_npi = sorted(set(combined_df['cand_npi'].dropna()))
print(f'Distinct cand_npi count: {len(distinct_cand_npi)}')
distinct_cand_npi
```
