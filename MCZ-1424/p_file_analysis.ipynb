```python
import pandas as pd
from IPython.display import display

# path = "/Users/tomcraig/dev/mcz_notebooks_and_queries/MCZ-1424/NPPES_Data_Dissemination_January_2026/p_file_chunk.csv"
path = "/Users/tomcraig/dev/mcz_notebooks_and_queries/MCZ-1424/NPPES_Data_Dissemination_January_2026/npidata_pfile_20050523-20260111.csv"

# Prefer fast parsing; use pyarrow if available and avoid dtype inference
read_kwargs = {"dtype": str}
try:
    import pyarrow  # noqa: F401
    read_kwargs["engine"] = "pyarrow"
except Exception:
    pass

# Read only the header first to list columns
# df_header = pd.read_csv(
#     path,
#     nrows=0,
#     usecols=lambda col: "taxonomy" in col.lower(),
#     **read_kwargs,
# )
# print(f"Columns ({len(df_header.columns)})")
# for c in df_header.columns:
#     print(c)

# Read a manageable sample for quick inspection
# sample_rows = 1000
# df_temp = pd.read_csv(path, nrows=sample_rows, **read_kwargs)
# df_temp = pd.read_csv(path, **read_kwargs)

# test NPI
npi = "1003000415"
chunk = None
chunk_size = 100000  # 100k rows at a time

# Read file in chunks
# Count total rows in file
total_rows = sum(1 for _ in open(path)) - 1  # subtract 1 for header
print(f"Total rows in file: {total_rows:,}\n")
chunks = []
rows_processed = 0

for i, chunk in enumerate(pd.read_csv(path, chunksize=chunk_size, **read_kwargs)):
    rows_processed += len(chunk)
    progress_pct = (rows_processed / total_rows) * 100
    
    # Search for NPI in this chunk
    matching_rows = chunk[chunk["NPI"].str.contains(npi, na=False)]
    
    if not matching_rows.empty:
        chunks.append(matching_rows)
        print(f"Chunk {i}: Found {len(matching_rows)} matches | Progress: {progress_pct:.1f}% ({rows_processed:,}/{total_rows:,})")
        break
    else:
        print(f"Chunk {i}: No matches | Progress: {progress_pct:.1f}% ({rows_processed:,}/{total_rows:,})")

# Combine all matching chunks
if chunks:
    chunk = pd.concat(chunks, ignore_index=True)
    print(f"\nTotal matches found: {len(chunk)}")
else:
    chunk = pd.DataFrame()
    print(f"\nNo matches found for NPI: {npi}")

# Print columns containing "taxonomy"
if not chunk.empty:
    taxonomy_cols = ["Provider First Name", "Provider Last Name (Legal Name)", "NPI"] + [col for col in chunk.columns if "taxonomy" in col.lower()]
    print(f"\nTaxonomy columns found: {taxonomy_cols}")
    display(chunk[taxonomy_cols])
```
