# Candidates Example: Column Analysis
This notebook loads `Candidates_example.csv` and displays all column names.

```python
import pandas as pd

csv_path = 'Candidates_example.csv'
df = pd.read_csv(csv_path)
print(f'Loaded {csv_path} with {len(df):,} rows and {len(df.columns)} columns.')
```

```python
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 0)

print(f'Total columns: {len(df.columns)}')
df.head(5)
# print('Column names:')
```

```python
npis_to_find = [
1063490852,
1194841254,
1790076586,
1336109925,
1457990731,
1518092048,
]

# Smaller dataframe with entries whose NPI is in npis_to_find.
npis_to_find_set = {str(npi).strip() for npi in npis_to_find}
matched_npi_df = df[df['NPI'].astype(str).str.strip().isin(npis_to_find_set)].copy()
print(f'Matched rows by NPI: {len(matched_npi_df):,}')

requested_columns = [
    'NPI',
    'degree',
    'educations',
    'experience',
    'certifications',
    'EducationTrainingLocation',
    'EducationTrainingDate',
    'EducationTrainingYears',
    'ResidencyTrainingLocation',
    'ResidencyTrainingDate',
    'ResidencyTrainingYears',
]

# Match requested names to actual dataframe columns, case-insensitively.
columns_lower_map = {c.lower(): c for c in matched_npi_df.columns}
selected_actual_columns = [columns_lower_map[c.lower()] for c in requested_columns if c.lower() in columns_lower_map]
missing_columns = [c for c in requested_columns if c.lower() not in columns_lower_map]

candidate_training_df = matched_npi_df[selected_actual_columns].head(100).copy()
candidate_training_df.columns = [
    next(req for req in requested_columns if req.lower() == actual.lower())
    for actual in selected_actual_columns
]

if missing_columns:
    print('Missing columns:', missing_columns)

candidate_training_df
```

```python
zocdoc_path = 'zocdoc data.xlsx'
zocdoc_df = pd.read_excel(zocdoc_path, dtype=str)
zocdoc_df = zocdoc_df[zocdoc_df['results'].ne('summary')].copy()

def normalize_npi_column(npi_column):
    return npi_column.astype('string').str.extract(r'(\d{10})', expand=False)

zocdoc_df['NormalizedNPI'] = normalize_npi_column(zocdoc_df['NPI'])
candidate_npis = normalize_npi_column(df['NPI'])

zocdoc_npi_set = set(zocdoc_df['NormalizedNPI'].dropna())
candidate_npi_set = set(candidate_npis.dropna())
matching_zocdoc_npi_set = zocdoc_npi_set.intersection(candidate_npi_set)

zocdoc_matches_df = zocdoc_df[
    zocdoc_df['NormalizedNPI'].isin(matching_zocdoc_npi_set)
].copy()

print(f'Zocdoc rows: {len(zocdoc_df):,}')
print(f'Zocdoc unique NPIs: {len(zocdoc_npi_set):,}')
print(f'Zocdoc unique NPIs in candidates df: {len(matching_zocdoc_npi_set):,}')
print(f'Zocdoc rows with NPIs in candidates df: {len(zocdoc_matches_df):,}')

zocdoc_matches_df.head(20)
```

```python
def has_non_empty_value(value):
    if isinstance(value, (list, tuple, set, dict)):
        return len(value) > 0
    if value is None or pd.isna(value):
        return False

    value_text = str(value).strip()
    return value_text.lower() not in {'', 'nan', 'null', 'none', '[]'}

zocdoc_matching_npis_df = zocdoc_df[
    zocdoc_df['NormalizedNPI'].isin(matching_zocdoc_npi_set)
].copy()

zocdoc_matches_with_education_df = zocdoc_matching_npis_df[
    zocdoc_matching_npis_df['Education'].map(has_non_empty_value)
].copy()
matching_zocdoc_npis_with_education_set = set(
    zocdoc_matches_with_education_df['NormalizedNPI'].dropna()
)

display_columns = [
    'NormalizedNPI',
    'NPI',
    'Name',
    'Education',
    'Specialty',
    'URL',
]

print(f'Matching Zocdoc rows: {len(zocdoc_matching_npis_df):,}')
print(
    'Matching Zocdoc rows with populated Education: '
    f'{len(zocdoc_matches_with_education_df):,}'
)
print(
    'Unique matching Zocdoc NPIs with populated Education: '
    f'{len(matching_zocdoc_npis_with_education_set):,}'
)

zocdoc_matches_with_education_df[display_columns].head(100)
```

```python
additional_npis_to_find = [
    1114059979,
    1174812614,
    1093713166,
    1790990943,
    1184120701,
    1205828449,
    1063437788,
    1104944891,
    1740406933,
    1356336614,
    1730421884,
    1366743056,
    1003870841,
    1285192609,
]

npis_to_find = list(dict.fromkeys([*npis_to_find, *additional_npis_to_find]))

npis_to_find_list = [str(npi).strip() for npi in npis_to_find]
npis_to_find_set = set(npis_to_find_list)
npi_order_df = pd.DataFrame(
    {
        'NormalizedNPI': npis_to_find_list,
        'NPIOrder': range(len(npis_to_find_list)),
    }
)

candidate_columns_lower_map = {column.lower(): column for column in df.columns}
candidate_actual_columns = [
    candidate_columns_lower_map[column.lower()]
    for column in requested_columns
    if column.lower() in candidate_columns_lower_map
]
candidate_column_name_map = {
    actual_column: f'Candidate_{requested_column}'
    for requested_column in requested_columns
    for actual_column in [candidate_columns_lower_map.get(requested_column.lower())]
    if actual_column is not None
}
missing_candidate_columns = [
    column
    for column in requested_columns
    if column.lower() not in candidate_columns_lower_map
]

candidate_lookup_df = df.copy()
candidate_lookup_df['NormalizedNPI'] = normalize_npi_column(candidate_lookup_df['NPI'])
candidate_npis_to_find_df = candidate_lookup_df[
    candidate_lookup_df['NormalizedNPI'].isin(npis_to_find_set)
][['NormalizedNPI', *candidate_actual_columns]].rename(
    columns=candidate_column_name_map
)

zocdoc_lookup_df = zocdoc_df.copy()
zocdoc_lookup_df['NormalizedNPI'] = normalize_npi_column(zocdoc_lookup_df['NPI'])
zocdoc_npis_to_find_df = zocdoc_lookup_df[
    zocdoc_lookup_df['NormalizedNPI'].isin(npis_to_find_set)
][['NormalizedNPI', 'NPI', 'Name', 'Education']].rename(
    columns={
        'NPI': 'Zocdoc_NPI',
        'Name': 'Zocdoc_Name',
        'Education': 'Zocdoc_Education',
    }
)

npi_education_comparison_df = npi_order_df.merge(
    candidate_npis_to_find_df,
    on='NormalizedNPI',
    how='left',
).merge(
    zocdoc_npis_to_find_df,
    on='NormalizedNPI',
    how='left',
).sort_values('NPIOrder').drop(columns=['NPIOrder'])

if missing_candidate_columns:
    print('Missing candidate columns:', missing_candidate_columns)

print(f'Candidate rows for npis_to_find: {len(candidate_npis_to_find_df):,}')
print(f'Zocdoc rows for npis_to_find: {len(zocdoc_npis_to_find_df):,}')
print(f'Comparison rows: {len(npi_education_comparison_df):,}')

with pd.option_context('display.max_colwidth', None):
    display(npi_education_comparison_df)
```
